```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/dymiyata/wva-ml-ai-2-2026-2027/refs/heads/main/classification_data/classification_class.csv")
```

```python
df.head()
```

### Scikit-learn

A library for datascience/basic machine learning.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score
```

```python
X = df.drop("target", axis=1)
y = df["target"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=2026
)
```

```python
model_log = LogisticRegression()
```

```python
model_log.fit(X_train, y_train)
```

```python
y_pred_train = model_log.predict(X_train)
y_pred_test = model_log.predict(X_test)
```

```python
print(confusion_matrix(y_train, y_pred_train))
print(accuracy_score(y_train, y_pred_train))
```

```python
print(confusion_matrix(y_test, y_pred_test))
print(accuracy_score(y_test, y_pred_test))
```

### Tensor Flow

Library for deep stuff.

```python
from tensorflow import keras
```

```python
model_neural = keras.Sequential([
    keras.layers.Input(shape=(10,)),
    keras.layers.Dense(200, activation="relu"),
    keras.layers.Dense(100, activation="relu"),
    keras.layers.Dense(1, activation="sigmoid")
])
```

```python
model_neural.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
model_neural.fit(X_train, y_train, epochs=16)
```

```python
y_pred_train_n = (model_neural.predict(X_train) > 0.5).astype(int)
y_pred_test_n = (model_neural.predict(X_test)>0.5).astype(int)
```

```python
print(confusion_matrix(y_train, y_pred_train_n))
print(accuracy_score(y_train, y_pred_train_n))


print(confusion_matrix(y_test, y_pred_test_n))
print(accuracy_score(y_test, y_pred_test_n))
```
