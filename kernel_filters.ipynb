````python
url = "https://colinmathers.com/wp-content/uploads/2014/01/about_erdos.jpg"
response = requests.get(url)
img = Image.open(BytesIO(response.content)).convert("L")
img = np.array(img)
````

````python
plt.imshow(img, cmap="gray")
````

````python
# kernel = np.array([
#     [-1, 0, 1],
#     [-1, 0, 1],
#     [-1, 0, 1]
# ])
kernel = np.array([
    [-1, -1, -1],
    [0, 0, 0],
    [1, 1, 1]
])
````

````python
filtered = convolve2d(img, kernel, mode="same")


plt.subplot(1,2,1)
plt.imshow(img, cmap="gray")

plt.subplot(1,2,2)
plt.imshow(filtered, cmap="gray")

plt.show()
````
